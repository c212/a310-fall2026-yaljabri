Question 2
ndmin sets the minimum number of dimensions. The function prints the number of dimensions and the array.

```python
def what(n):
    a = np.array([1, 2, 3, 4], ndmin=n)
    print(a.ndim)
    return a

print(what(1))
print(what(2))
print(what(3))
```

Question 17
A view shares underlying data, so its .base points to the original array. Copies own their data.

```python
a = np.array([1, 2, 3, 4, 5])
d = a.view()

print(a.base)
print(d.base)

b = a
print(b.base)

c = a.copy()
print(c.base)
```

Question 24
Calculate after-tax incomes and use Boolean indexing to find the person with the highest income. Salaries are in thousands of dollars. The answer is Tim, $81,000.

```python
alice = [99, 101, 103]
bob = [110, 108, 105]
tim = [90, 88, 85]

salaries = np.array([alice, bob, tim])

taxation = np.array([[0.2, 0.25, 0.22],
                     [0.4, 0.5, 0.5],
                     [0.1, 0.2, 0.1]])

names = np.array(["Alice", "Bob", "Tim"])

income = salaries - salaries * taxation

highest = np.max(income)

person = names[np.any(income == highest, axis=1)]

print(person)
print(highest)
```
