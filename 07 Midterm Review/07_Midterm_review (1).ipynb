Question 1 — Short definitions

1. Supervised Machine Learning: A type of machine learning where the computer learns from labeled data to predict an answer.

2. Unsupervised Machine Learning: A type of machine learning where the computer finds patterns in data without labeled answers.

3. Linear Regression: A method used to predict a number by finding a straight-line relationship between variables.

4. Logistic Regression: A method used to predict a category, usually using probabilities, such as yes or no.

5. K-Means Clustering: An algorithm that divides data into K groups based on similarities.

6. K-Nearest Neighbors : An algorithm that predicts an answer by looking at the K closest data points.

7. Artificial Neural Network: A machine learning model made of connected neurons that learn patterns from data.

8. Basic Statistics:
- Average: The sum of all values divided by how many values there are.
- Variance: Measures how spread out values are from the average.
- Standard Deviation: The square root of variance.
9. Support Vector Machine : A machine learning algorithm that finds the best boundary to separate different classes of data.

10. Regular Expression: A pattern used to search, match, or replace specific text in strings.

Question 4 — Coin flips (a–d)

```python
import random

N = 10000

def probability(part):
    count = 0
    for i in range(N):
        flips = "".join(random.choice("HT") for j in range(100))
        tails = "TTTTTT" in flips
        heads = "HHHHHH" in flips
        if part == "a":
            good = tails
        elif part == "b":
            good = tails and heads
        elif part == "c":
            good = tails or heads
        else:
            good = tails and not heads
        if good:
            count += 1
    return count / N

for part in "abcd":
    print(part, probability(part))
```

Question 5 — LinkedList

```python
class LinkedList:
    def __init__(self, value, next):
        self.value = value
        self.next = next

    def insert(self, value):
        return LinkedList(value, self)

    def show(self):
        if self.next is None:
            return str(self.value)
        return str(self.value) + " " + self.next.show()

    def find(self, value):
        if self.value == value:
            return True
        if self.next is None:
            return False
        return self.next.find(value)

    def remove(self, value):
        if self.value == value:
            return self.next
        if self.next is not None:
            self.next = self.next.remove(value)
        return self

a = LinkedList(5, None)
a = a.insert(4)
a = a.insert(3)
a = a.insert(2)
a = a.insert(1)
print(a.show())
print(a.find(3))
a = a.remove(3)
print(a.show())
```

Last question
 — Recursive sorting with longest, rest, merge, **sort**

```python
def longest(seq):
    if not seq:
        return []
    i = 1
    while i < len(seq) and seq[i] >= seq[i - 1]:
        i += 1
    return seq[:i]

def rest(seq):
    return seq[len(longest(seq)):]

def merge(a, b):
    result = []
    i = 0
    j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    return result + a[i:] + b[j:]

def sort(seq):
    if len(seq) <= 1:
        return seq
    return merge(longest(seq), sort(rest(seq)))

numbers = [4, 5, 2, 3, 1, 7, 6]
print(longest(numbers))
print(rest(numbers))
print(sort(numbers))
```
