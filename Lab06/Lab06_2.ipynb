Build a weighted sample space
For ten fair coin flips, comb(10, heads) counts how many possible sequences contain a particular number of heads. There are 2**10 equally likely sequences.

```python
import matplotlib.pyplot as plt
from math import comb

flips = 10
weighted_sample_space = {heads: comb(flips, heads) for heads in range(flips + 1)}
print(weighted_sample_space)
```

Plot the number of combinations
The middle head counts have more possible combinations.

```python
heads = list(weighted_sample_space.keys())
counts = list(weighted_sample_space.values())
plt.scatter(heads, counts)
plt.xlabel("Head-count")
plt.ylabel("Number of coin-flip combinations")
plt.title("Combinations from 10 coin flips")
plt.show()
```

Plot the probabilities
Divide each combination count by the total number of outcomes to obtain the probability.

```python
total = sum(counts)
probabilities = [count / total for count in counts]
plt.plot(heads, probabilities, marker="o")
plt.xlabel("Head-count")
plt.ylabel("Probability")
plt.title("Probability distribution for 10 fair coin flips")
plt.show()
```

Highlight an extreme outcome
We shade the head counts from 8 through 10 and calculate their exact probability.

```python
plt.plot(heads, probabilities, marker="o")
plt.fill_between(heads, probabilities, where=[h >= 8 for h in heads], alpha=0.4)
plt.xlabel("Head-count")
plt.ylabel("Probability")
plt.title("Extreme outcomes: at least 8 heads")
plt.show()
print("P(at least 8 heads):", sum(p for h, p in zip(heads, probabilities) if h >= 8))
```

Answer
We create a weighted sample space, divide each count by the total number of possible coin-flip sequences, and use Matplotlib to graph head-count against probability. For ten flips, five heads is the most likely exact count. The probability of eight or more heads is 0.0546875.
