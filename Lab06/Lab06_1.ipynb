Line plot
A line plot connects points and helps us see how a value changes.

```python
import matplotlib.pyplot as plt

x = list(range(10))
y = [2 * value for value in x]
plt.plot(x, y)
plt.title("Line plot: y = 2x")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Scatter plot
A scatter plot shows each data point separately

```python
plt.scatter(x, y)
plt.title("Scatter plot: y = 2x")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Shading an interval
fill_between highlights the area under the line for a selected range. Here we shade from x = 2 to x = 6.

```python
plt.plot(x, y)
plt.fill_between(x, y, where=[2 <= value <= 6 for value in x], alpha=0.4)
plt.title("Shaded interval from 2 to 6")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

In Chapter 2, we use line plots, scatter plots, and shaded areas with plt.fill_between(). Line plots show a trend, scatter plots show individual values, and shaded regions highlight intervals.
