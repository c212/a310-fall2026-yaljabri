Midterm LinkedList

```python

class LinkedList:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next

    def insert(self, value):
        return LinkedList(value, self)

    def show(self):
        if self.next is None:
            return str(self.value)
        return str(self.value) + " " + self.next.show()


a = LinkedList(5, None)
print(a.show())

a = a.insert(4)
print(a.show())

a = a.insert(3)
print(a.show())

a = a.insert(2)
print(a.show())

a = a.insert(1)
print(a.show())
```
