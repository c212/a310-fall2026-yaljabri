```python

# Problem 1 - Towers of Hanoi

def hanoi(n, start, end, middle):
    if n == 1:
        print("Move disk 1 from", start, "to", end)
        return

    hanoi(n - 1, start, middle, end)

    print("Move disk", n, "from", start, "to", end)

    hanoi(n - 1, middle, end, start)


hanoi(3, "A", "C", "B")
```

Problem 1: Towers of Hanoi

I used a recursive function to solve the Towers of Hanoi problem. The function moves the smaller disks first so that the larger disk can move to the correct tower. It repeats this process until all the disks are moved from Tower A to Tower C. For three disks, the program takes seven moves.

```python

# Problem 2 - QuickSort

def quicksort(numbers):
    if len(numbers) <= 1:
        return numbers

    pivot = numbers[0]

    left = []
    right = []

    for number in numbers[1:]:
        if number < pivot:
            left.append(number)
        else:
            right.append(number)

    return quicksort(left) + [pivot] + quicksort(right)



numbers = [8, 3, 1, 7, 0, 10, 2]

print("Original list:", numbers)
print("Sorted list:", quicksort(numbers))
```

Problem 2: QuickSort

I created a QuickSort function to sort a list of numbers from smallest to largest. The program chooses the first number as the pivot. It puts smaller numbers on the left and larger numbers on the right. Then it uses recursion to sort both sides and combines them into one sorted list.

```python

# Problem 3 - Pig Latin using Regular Expressions

import re

def pig_latin(word):

    match = re.fullmatch(r"([^aeiouAEIOU]+)([aeiouAEIOU].*)", word)

    if match:
        return match.group(2) + match.group(1) + "ay"


    if re.fullmatch(r"[aeiouAEIOU].*", word):
        return word + "way"

    return word


def translate(text):

    return re.sub(
        r"[A-Za-z]+",
        lambda match: pig_latin(match.group()),
        text
    )


text = """Hi I am here now.

What is going on?

Are you having a reasonable good Friday?"""

print(translate(text))
```

Problem 3: Pig Latin

I wrote a Python program that uses regular expressions to change English words into Pig Latin. If a word starts with consonants, the program moves them to the end and adds "ay". If a word starts with a vowel, it adds "way". I used the `re` module to find words in a sentence and convert them while keeping the punctuation.
